```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from scipy import stats
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans, AgglomerativeClustering
from sklearn.metrics import silhouette_score, adjusted_rand_score
from sklearn.decomposition import PCA
from scipy.cluster.hierarchy import linkage, dendrogram, fcluster

#Loading the dataset
url = "https://raw.githubusercontent.com/salemprakash/EDA/main/Data/HepatitisCdata.csv"
hcv_data = pd.read_csv(url)

#Dropping the index column and filling missing lab values with the median
hcv_data = hcv_data.drop(columns=["Unnamed: 0"])

lab_features = ["ALB", "ALP", "ALT", "AST", "BIL", "CHE", "CHOL", "CREA", "GGT", "PROT"]
numeric_features = ["Age"] + lab_features

for column in lab_features:
    hcv_data[column] = hcv_data[column].fillna(hcv_data[column].median())

#Simplified target: healthy donors vs liver disease patients
hcv_data["Diagnosis"] = np.where(
    hcv_data["Category"].str.startswith("0"),
    "Healthy",
    "Disease"
)

hcv_data["Age_Group"] = pd.cut(
    hcv_data["Age"],
    bins=[0, 40, 50, 60, 100],
    labels=["<40", "40-50", "50-60", "60+"]
)

print(hcv_data.shape)
print(hcv_data.isnull().sum().sum(), "missing values left")
```

```python
#1D Statistical Analysis (Univariate)
one_d_stats = pd.DataFrame({
    "mean": hcv_data[numeric_features].mean(),
    "median": hcv_data[numeric_features].median(),
    "mode": hcv_data[numeric_features].mode().iloc[0],
    "variance": hcv_data[numeric_features].var(),
    "std_dev": hcv_data[numeric_features].std(),
    "range": hcv_data[numeric_features].max() - hcv_data[numeric_features].min(),
    "Q1": hcv_data[numeric_features].quantile(0.25),
    "Q3": hcv_data[numeric_features].quantile(0.75),
    "IQR": hcv_data[numeric_features].quantile(0.75) - hcv_data[numeric_features].quantile(0.25),
    "skewness": hcv_data[numeric_features].skew(),
    "kurtosis": hcv_data[numeric_features].kurt(),
    "coeff_of_variation": hcv_data[numeric_features].std() / hcv_data[numeric_features].mean()
})

print(one_d_stats.round(3))
```

```python
#Outlier detection using the IQR rule
for column in numeric_features:
    q1 = hcv_data[column].quantile(0.25)
    q3 = hcv_data[column].quantile(0.75)
    iqr = q3 - q1
    outliers = hcv_data[
        (hcv_data[column] < q1 - 1.5 * iqr) |
        (hcv_data[column] > q3 + 1.5 * iqr)
    ]
    print(column, "-> outliers:", len(outliers))

#Frequency and relative frequency of categorical columns
for column in ["Category", "Sex", "Diagnosis", "Age_Group"]:
    frequency_table = pd.DataFrame({
        "count": hcv_data[column].value_counts(),
        "percent": hcv_data[column].value_counts(normalize=True).mul(100).round(2)
    })
    print("\nFrequency table for", column)
    print(frequency_table)

#Normality test (Shapiro-Wilk) for every numerical column
for column in numeric_features:
    statistic, p_value = stats.shapiro(hcv_data[column])
    print(column, "-> W =", round(statistic, 4), ", p-value =", p_value)
```

```python
#1D visualization: histograms of all lab values
hcv_data[lab_features].hist(bins=30, figsize=(15, 10))
plt.suptitle("Distribution of Lab Values")
plt.tight_layout()
plt.show()

#Box plots of scaled lab values to compare spread and outliers
plt.figure(figsize=(12, 5))

sns.boxplot(
    data=(hcv_data[lab_features] - hcv_data[lab_features].mean()) / hcv_data[lab_features].std()
)

plt.title("Standardised Lab Values")
plt.show()

#Pie chart of diagnosis category
hcv_data["Category"].value_counts().plot(
    kind="pie",
    autopct="%1.1f%%",
    figsize=(7, 7)
)

plt.title("Category Proportion")
plt.ylabel("")
plt.show()
```

```python
#2D Statistical Analysis (Bivariate)
#Covariance matrix
print("Covariance matrix:")
print(hcv_data[numeric_features].cov().round(2))

#Pearson and Spearman correlation between AST and GGT
pearson_r, pearson_p = stats.pearsonr(hcv_data["AST"], hcv_data["GGT"])
spearman_r, spearman_p = stats.spearmanr(hcv_data["AST"], hcv_data["GGT"])

print("\nPearson  (AST, GGT): r =", round(pearson_r, 4), ", p =", pearson_p)
print("Spearman (AST, GGT): rho =", round(spearman_r, 4), ", p =", spearman_p)

#Pearson and Spearman correlation heatmaps
fig, axes = plt.subplots(1, 2, figsize=(18, 7))

sns.heatmap(
    hcv_data[numeric_features].corr(method="pearson"),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    ax=axes[0]
)
axes[0].set_title("Pearson Correlation Matrix")

sns.heatmap(
    hcv_data[numeric_features].corr(method="spearman"),
    annot=True,
    fmt=".2f",
    cmap="coolwarm",
    ax=axes[1]
)
axes[1].set_title("Spearman Correlation Matrix")

plt.show()
```

```python
#Numerical vs categorical: group statistics
print(hcv_data.groupby("Diagnosis")[numeric_features].agg(["mean", "median"]).round(2).T)

#Mann-Whitney U test (lab values are skewed): Healthy vs Disease
for column in lab_features:
    healthy = hcv_data[hcv_data["Diagnosis"] == "Healthy"][column]
    disease = hcv_data[hcv_data["Diagnosis"] == "Disease"][column]
    u_stat, u_p = stats.mannwhitneyu(healthy, disease)
    print(column, "-> U =", round(u_stat, 1), ", p =", u_p)

#Welch t-test: does ALB differ between Healthy and Disease?
t_stat, t_p = stats.ttest_ind(
    hcv_data[hcv_data["Diagnosis"] == "Healthy"]["ALB"],
    hcv_data[hcv_data["Diagnosis"] == "Disease"]["ALB"],
    equal_var=False
)
print("\nWelch t-test (ALB by Diagnosis): t =", round(t_stat, 3), ", p =", t_p)

#One-way ANOVA and Kruskal-Wallis: does AST differ across the disease categories?
category_groups = [group["AST"].values for _, group in hcv_data.groupby("Category")]
f_stat, f_p = stats.f_oneway(*category_groups)
h_stat, h_p = stats.kruskal(*category_groups)
print("ANOVA (AST by Category): F =", round(f_stat, 3), ", p =", f_p)
print("Kruskal-Wallis (AST by Category): H =", round(h_stat, 3), ", p =", h_p)

#Categorical vs categorical: contingency table and chi-square test
contingency_table = pd.crosstab(hcv_data["Sex"], hcv_data["Diagnosis"])
print("\nContingency table (Sex vs Diagnosis):")
print(contingency_table)

chi2, chi_p, dof, expected = stats.chi2_contingency(contingency_table)
print("\nChi-square (Sex vs Diagnosis): chi2 =", round(chi2, 3), ", dof =", dof, ", p =", chi_p)
```

```python
#Scatter plot with regression line: AST vs GGT
plt.figure(figsize=(8, 6))

sns.regplot(
    x="AST",
    y="GGT",
    data=hcv_data,
    scatter_kws={"alpha": 0.5}
)

plt.title("AST vs GGT")
plt.show()

#Box plot of AST across categories
plt.figure(figsize=(10, 5))

sns.boxplot(
    x="Category",
    y="AST",
    data=hcv_data
)

plt.xticks(rotation=20)
plt.title("AST by Category")
plt.show()

#Stacked bar chart of diagnosis by sex
pd.crosstab(hcv_data["Sex"], hcv_data["Diagnosis"], normalize="index").plot(
    kind="bar",
    stacked=True,
    figsize=(7, 5)
)

plt.title("Diagnosis Proportion by Sex")
plt.ylabel("Proportion")
plt.show()
```

```python
#3D Statistical Analysis (Trivariate)
#Three-way pivot table: mean AST by age group, sex and diagnosis
pivot_3d = pd.pivot_table(
    hcv_data,
    values="AST",
    index=["Age_Group", "Sex"],
    columns="Diagnosis",
    aggfunc="mean",
    observed=True
)

print(pivot_3d.round(2))

#Three-way contingency table: age group x sex x diagnosis
print(pd.crosstab(
    [hcv_data["Age_Group"], hcv_data["Sex"]],
    hcv_data["Diagnosis"],
    margins=True
))

#Partial correlation between AST and GGT, controlling for ALT
def partial_correlation(data, x, y, z):
    r_xy = data[x].corr(data[y])
    r_xz = data[x].corr(data[z])
    r_yz = data[y].corr(data[z])
    return (r_xy - r_xz * r_yz) / np.sqrt((1 - r_xz ** 2) * (1 - r_yz ** 2))

print("\nPartial correlation (AST, GGT | ALT):",
      round(partial_correlation(hcv_data, "AST", "GGT", "ALT"), 4))

#Grouped statistics across three variables: mean lab values by Category and Sex
print(hcv_data.groupby(["Category", "Sex"])[["ALB", "AST", "GGT", "CHE"]].mean().round(2))
```

```python
#3D scatter plot: AST, GGT and ALB coloured by category
fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection="3d")

for category in hcv_data["Category"].unique():
    subset = hcv_data[hcv_data["Category"] == category]
    ax.scatter(
        subset["AST"],
        subset["GGT"],
        subset["ALB"],
        label=category,
        alpha=0.6,
        s=15
    )

ax.set_xlabel("AST")
ax.set_ylabel("GGT")
ax.set_zlabel("ALB")
ax.set_title("3D Scatter: AST, GGT, ALB")
ax.legend()
plt.show()

#Heatmap of mean AST across age group and sex
plt.figure(figsize=(7, 5))

sns.heatmap(
    pd.pivot_table(
        hcv_data,
        values="AST",
        index="Age_Group",
        columns="Sex",
        aggfunc="mean",
        observed=True
    ),
    annot=True,
    fmt=".1f",
    cmap="YlOrRd"
)

plt.title("Mean AST by Age Group and Sex")
plt.show()

#Pair plot of three lab values with diagnosis hue
sns.pairplot(
    hcv_data[["AST", "GGT", "ALB", "Diagnosis"]],
    hue="Diagnosis",
    plot_kws={"alpha": 0.5, "s": 15}
)

plt.show()
```

```python
#K-Means Clustering
#Lab values are highly skewed, so log-transform them before scaling
cluster_data = np.log1p(hcv_data[lab_features])

scaler = StandardScaler()
scaled_data = scaler.fit_transform(cluster_data)

#Elbow method and silhouette score to choose k
inertia_values = []
silhouette_values = []
k_range = range(2, 11)

for k in k_range:
    kmeans_model = KMeans(n_clusters=k, n_init=10, random_state=42)
    kmeans_labels = kmeans_model.fit_predict(scaled_data)
    inertia_values.append(kmeans_model.inertia_)
    silhouette_values.append(silhouette_score(scaled_data, kmeans_labels))

fig, axes = plt.subplots(1, 2, figsize=(14, 5))

axes[0].plot(k_range, inertia_values, marker="o")
axes[0].set_title("Elbow Method")
axes[0].set_xlabel("Number of clusters (k)")
axes[0].set_ylabel("Inertia")

axes[1].plot(k_range, silhouette_values, marker="o", color="tab:green")
axes[1].set_title("Silhouette Score")
axes[1].set_xlabel("Number of clusters (k)")
axes[1].set_ylabel("Silhouette")

plt.show()
```

```python
#Fitting the final K-Means model
optimal_k = 3

kmeans = KMeans(n_clusters=optimal_k, n_init=10, random_state=42)
hcv_data["KMeans_Cluster"] = kmeans.fit_predict(scaled_data)

print("Cluster sizes:")
print(hcv_data["KMeans_Cluster"].value_counts().sort_index())

print("\nCluster profiles (mean of original lab values):")
print(hcv_data.groupby("KMeans_Cluster")[numeric_features].mean().round(2))

#Comparing clusters with the actual categories
print("\nClusters vs Category:")
print(pd.crosstab(hcv_data["KMeans_Cluster"], hcv_data["Category"]))

print("\nAdjusted Rand Index (K-Means vs Diagnosis):",
      round(adjusted_rand_score(hcv_data["Diagnosis"], hcv_data["KMeans_Cluster"]), 4))

#Visualising the clusters in 2D using PCA
pca = PCA(n_components=2)
pca_components = pca.fit_transform(scaled_data)

print("\nExplained variance ratio:", pca.explained_variance_ratio_.round(3))

fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.scatterplot(
    x=pca_components[:, 0],
    y=pca_components[:, 1],
    hue=hcv_data["KMeans_Cluster"],
    palette="tab10",
    s=25,
    ax=axes[0]
)

centroids_pca = pca.transform(kmeans.cluster_centers_)
axes[0].scatter(
    centroids_pca[:, 0],
    centroids_pca[:, 1],
    c="black",
    marker="X",
    s=200,
    label="Centroids"
)
axes[0].set_title("K-Means Clusters (PCA projection)")
axes[0].legend()

sns.scatterplot(
    x=pca_components[:, 0],
    y=pca_components[:, 1],
    hue=hcv_data["Category"],
    s=25,
    ax=axes[1]
)
axes[1].set_title("Actual Categories (PCA projection)")

plt.show()
```

```python
#Hierarchical Clustering
#Dendrogram using Ward linkage (dataset is small, so all rows are used)
linkage_matrix = linkage(scaled_data, method="ward")

plt.figure(figsize=(14, 6))

dendrogram(
    linkage_matrix,
    truncate_mode="lastp",
    p=30,
    leaf_rotation=90,
    show_contracted=True
)

plt.title("Hierarchical Clustering Dendrogram (Ward)")
plt.xlabel("Cluster size")
plt.ylabel("Distance")
plt.show()

#Comparing linkage methods using silhouette score
for method in ["ward", "complete", "average", "single"]:
    agglomerative = AgglomerativeClustering(n_clusters=optimal_k, linkage=method)
    labels = agglomerative.fit_predict(scaled_data)
    print(method, "linkage -> silhouette =", round(silhouette_score(scaled_data, labels), 4))
```

```python
#Final hierarchical model with Ward linkage
hcv_data["Hierarchical_Cluster"] = fcluster(linkage_matrix, t=optimal_k, criterion="maxclust") - 1

print("Hierarchical cluster sizes:")
print(hcv_data["Hierarchical_Cluster"].value_counts().sort_index())

print("\nHierarchical cluster profiles:")
print(hcv_data.groupby("Hierarchical_Cluster")[numeric_features].mean().round(2))

print("\nHierarchical clusters vs Category:")
print(pd.crosstab(hcv_data["Hierarchical_Cluster"], hcv_data["Category"]))

#Agreement between the two clustering methods and with the true diagnosis
print("\nAdjusted Rand Index (K-Means vs Hierarchical):",
      round(adjusted_rand_score(hcv_data["KMeans_Cluster"], hcv_data["Hierarchical_Cluster"]), 4))
print("Adjusted Rand Index (Hierarchical vs Diagnosis):",
      round(adjusted_rand_score(hcv_data["Diagnosis"], hcv_data["Hierarchical_Cluster"]), 4))

#Side-by-side PCA view of both clusterings
fig, axes = plt.subplots(1, 2, figsize=(16, 6))

sns.scatterplot(
    x=pca_components[:, 0],
    y=pca_components[:, 1],
    hue=hcv_data["KMeans_Cluster"],
    palette="tab10",
    s=25,
    ax=axes[0]
)
axes[0].set_title("K-Means")

sns.scatterplot(
    x=pca_components[:, 0],
    y=pca_components[:, 1],
    hue=hcv_data["Hierarchical_Cluster"],
    palette="tab10",
    s=25,
    ax=axes[1]
)
axes[1].set_title("Hierarchical - Ward")

plt.show()
```
